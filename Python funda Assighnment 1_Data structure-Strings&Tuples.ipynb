## 1. String Concatenation: 
Write a Python program that takes two strings i.e string 1 “Hello ”, string 2 get name as 
input from the user and concatenates them together. Display the concatenated string as 
the output. 

```python
name = input("Enter your Name: ")
greeting = "Hello " + name
print(greeting)
```

```python
full_string = greeting + ", welcome to Python programming"
print(full_string)
```

## 2. String Slicing and Indexing: 
Write a Python program using the above concatenated string as input and performs the 
following tasks:  

```python
s = full_string   # "Hello Zara, welcome to Python programming"
print("First character:", s[0]) #a. Print the first character of the string.
```

```python
print("last character:",s[-1]) # b. Print the last character of the string.
```

```python
print("First 5 characters:", s[:5]) # c. Print the first 5 characters of the string.
```

```python
print("Last 11 characters:", s[-11:]) # d. Print the last 11 characters of the string.
```

```python
print("Reversed string:", s[::-1]) # e. Print the string in reverse.
```

### f. Use slicing and print the word “Python” from the existing string. 

```python
start = s.index("Python")
python_word = s[start:start + len("Python")]
print("Word Python:", python_word)
```

## Tuples (Creation, Modification and Access) :  
  
 
  




```python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60) # Create 1st tuple with values -> (10, 20, 30), 2nd tuple with values -> (40, 50, 60):
```

```python
t_combine = tuple1 + tuple2
print("Combined tuple:", t_combine) # a. Concatenate the two tuples and store it in “t_combine” 
```

```python
repeated = t_combine * 3
print("Repeated 3 times:", repeated) # b. Repeat the elements of “t_combine” 3 times
```

```python
print("Third element:", t_combine[2]) # c. Access the 3rd element from “t_combine”  
```

```python
print("First three elements:", t_combine[:3]) # d. Access the first three elements from “t_combine”  
```

```python
print("Last three elements:", t_combine[-3:]) #e. Access the last three elements from “t_combine”
```
